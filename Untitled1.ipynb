```python
import math

def q(a,b,n):
    for y in range(-a, a):
        x = int(b * math.sqrt(1 - (y**2) / (a**2)))
        print(" " * (b - x) + "*" * (2*x + 1))
    for x in range(n):
        print (" " * (b-3) + "|" * 5)

q(12,18,9)
```
